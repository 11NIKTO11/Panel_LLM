```python
def explore_dataset_values(df):
    """
    Analyzes a DataFrame to print the unique values for specific columns
    used in the processing functions.

    This helps in understanding the range of possible values for each raw field
    before it gets cleaned and standardized.

    Args:
        df (pd.DataFrame): The input DataFrame containing the raw survey data.
    """
    print("--- Starting Data Exploration: Unique Column Values ---\n")

    # --- 1. Define the columns to explore ---
    # List of columns with fixed names
    base_columns_to_check = [
        'GENDER', 'AGE1', 'EDU', 'KRAJ', 'OKRES', 'VMB',
        'INCOMEP', 'Q19', 'Q20', 'Q18', 'Q17', 'Q23'
    ]

    # Dynamically find all employment-related columns
    # This assumes the pattern 'EMPLOYMENT_' from your previous function.
    # If the original pattern was 'EMPL', you should use that here.
    # I'll use 'EMPL' to match the very first function you provided.
    employment_columns = [col for col in df.columns if 'EMPLOYMENT_' in col]

    # Combine the lists for a full exploration
    all_columns_to_explore = base_columns_to_check + employment_columns

    # --- 2. Iterate and print unique values ---
    for column_name in all_columns_to_explore:
        # First, check if the column actually exists in the DataFrame
        if column_name in df.columns:
            print(f"--- Column: '{column_name}' ---")
            try:
                # Get unique values and convert to a list for clean printing
                unique_values = df[column_name].unique().tolist()
                unique_values.sort()
                print(unique_values)
            except Exception as e:
                print(f"Could not retrieve unique values. Error: {e}")
            print("\n") # Add a newline for better readability
        else:
            # Print a warning if a column is not found
            print(f"--- Column: '{column_name}' ---")
            print("WARNING: This column was not found in the DataFrame.\n")

    print("--- Data Exploration Complete ---")

explore_dataset_values(df)
```

```python

# 11. Explore unique values in a specific column (if you have categorical data)
# Replace 'your_categorical_column' with an actual column name from your dataset.
# .value_counts() returns a series containing counts of unique values.
# This is a good way to understand the distribution of a categorical variable.
# We will check if there is a categorical column first.
categorical_cols = df.select_dtypes(include=['object', 'category']).columns
if len(categorical_cols) > 0:
    column_to_check = categorical_cols[0]
    print(f"\nValue counts for column '{column_to_check}':")
    print(df[column_to_check].value_counts())
else:
    print("\nNo categorical columns found to run value_counts() on.")

# 12. Basic Data Visualization
# A histogram can help you understand the distribution of your numerical data.
# We will select only numerical columns to plot.
numerical_cols = df.select_dtypes(include=['number']).columns
if len(numerical_cols) > 0:
    print("\nGenerating histograms for numerical columns...")
    df[numerical_cols].hist(bins=15, figsize=(15, 6), layout=(2, 4))
    plt.tight_layout()
    plt.show()
else:
    print("\nNo numerical columns to generate histograms for.")

# A pairplot can show relationships between variables.
# Note: This can be slow on very large datasets.
if len(numerical_cols) > 1:
    print("\nGenerating pairplot to see relationships between numerical variables...")
    sns.pairplot(df[numerical_cols].dropna())
    plt.show()
else:
    print("\nNot enough numerical columns to generate a pairplot.")
```
