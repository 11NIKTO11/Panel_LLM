```python
def explore_dataset_values(data:pd.DataFrame, columns:list[str]):
    print("--- Starting Data Exploration: Unique Column Values ---\n")

    for column_name in columns:
        # First, check if the column actually exists in the DataFrame
        if column_name in data.columns:
            print(f"--- Column: '{column_name}' ---")
            try:
                # Get unique values and convert to a list for clean printing
                unique_values = data[column_name].unique().tolist()
                unique_values.sort()
                print(unique_values)
            except Exception as e:
                print(f"Could not retrieve unique values. Error: {e}")
            print("\n") # Add a newline for better readability
        else:
            # Print a warning if a column is not found
            print(f"--- Column: '{column_name}' ---")
            print("WARNING: This column was not found in the DataFrame.\n")

    print("--- Data Exploration Complete ---")
```

```python
base_cols = ['GENDER', 'AGE1', 'EDU', 'KRAJ', 'OKRES', 'VMB', 'INCOMEP']
employment_cols = [col for col in df.columns if 'EMPLOYMENT_' in col]
question_cols = ['Q19', 'Q20', 'Q18', 'Q17', 'Q23']
columns_to_explore = base_cols + employment_cols + question_cols

explore_dataset_values(df, columns_to_explore)
```

```python
cols_to_plot = base_cols

# Plotting: create one figure per column for clarity
for col in cols_to_plot:
    series = df[col]
    # Drop missing entries
    series = series.dropna().astype(str)

    plt.figure(figsize=(8, 5))
    plt.title(f"Histogram of {col}")

    value_counts = series.astype(str).value_counts()
    value_counts.sort_index(inplace=True)
    sns.barplot(x=value_counts.index, y=value_counts.values)
    plt.xticks(rotation=45, ha='right',)
    plt.xlabel(col)
    plt.ylabel("Count")

    plt.tight_layout()
    plt.show()
```
