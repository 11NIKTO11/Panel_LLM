# Compare

Several saved runs of one topic beside each other, one block per comparison. The first cell is the only setup: name the topic and the runs (each by any text only its stem contains), the name the written files carry, whether anything is written, and whether figures use the topic's `plot.json`. `python compare.py <topic> <run> <run> ...` runs the same blocks in one go; `RUN_NAMES` may also be a mapping, label -> text, and the runs are then called by those labels; `load_comparisons(paths.CONFIG_DIR / "comparisons.json")["haiku"].runs` is such a mapping, a named set. `python analyze.py <topic>` lists a topic's runs. `PREDICTORS` names the columns the effects block tests (a list); `None` takes the dataset's grouping columns and its claims column. `DELTA` is what a zero share is replaced by before the log-ratio transform in that block. `WEIGHTED` weights each answer's option shares by its probability of voting when they are summed into a predicted distribution; `False` counts every answer the same.

```python
%matplotlib inline
from panel import paths
from panel.analysis import Output
from panel.analysis.comparison import areas, compass, describe, distances, effects, load_runs, shares, spreads
from panel.run import ResultStore

TOPIC = "election_2021"
RUN_NAMES = ["sod_default_cs_default_AnthropicLegacy_claude-haiku-4-5", "sod_default_cs_default_Anthropic_claude-haiku-5-5"]
NAME = TOPIC
SAVE = False
PREDICTORS = None
DELTA = 0.005
STYLED = True
WEIGHTED = True

STORE = ResultStore(paths.RESULTS_DIR, TOPIC)
RUNS = load_runs(STORE, RUN_NAMES, styled=STYLED, weighted=WEIGHTED)
OUT = Output(STORE, NAME, save=SAVE)
print(describe(RUNS))
```

## Shares side by side

The actual result, what the subjects claimed, and each run's prediction over all its subjects; the published polls are in the table only. A run's numbers and whiskers are those of its own country-level analysis.

```python
table, figures = shares(RUNS, OUT)
display(table.round(3), *figures)
```

## Distances

Every source against the actual result and against what the subjects claimed: the polls first, then the claimed distribution, then the runs. A poll is a published number: no interval, no turnout.

```python
table, figures = distances(RUNS, OUT)
display(table.round(4), *figures)
```

## By area

MAE and RMSE from the actual result in each area, for every run and for what the subjects claimed: a heatmap per metric, areas by sources, each cell with its interval underneath.

```python
table, figures = areas(RUNS, OUT)
display(table.round(3), *figures)
```

## Spread within an area

The standard deviation of each option's share over the subjects of one area, a panel per run: how far apart a run puts people who live in the same place.

```python
table, figures = spreads(RUNS, OUT)
display(table.round(3), *figures)
```

## What each run's answers follow

The MANOVA and PERMANOVA effect of each column, per run, as three heatmaps, columns by runs: Pillai's trace with its p-value, partial eta squared, and the PERMANOVA R squared with its p-value.

```python
table, figures = effects(RUNS, OUT, PREDICTORS, delta=DELTA)
display(table.round(3), *figures)
```

## Each run's map of the options against the experts'

One map per run, and per run the stress of its map, the Procrustes correlation and the Spearman correlation of its distances with the experts', each with an exact permutation p-value.

```python
table, figures = compass(RUNS, OUT)
display(table.round(3), *figures)
```
